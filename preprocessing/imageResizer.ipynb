```python
import os
import shutil
import cv2
```

```python
rgb_source_folder = "images_final"
source_folder = "dpt_final/"
destination_folder = "test_resize/"
```

```python
# rbg_folder_list = list(os.listdir(rgb_source_folder))
# rbg_folder_list.sort(key=lambda x: int(x.split(".")[0]))
source_folder_list = list(os.listdir(source_folder))
source_folder_list.sort(key=lambda x: int(x.split(".")[0]))
source_folder_list[:10]
```

```python
for img in source_folder_list:
    rgbImage = cv2.imread(f"{rgb_source_folder}{img}")
    rgbHeight, rgbWidth ,_ = rgbImage.shape
    path = f"{source_folder}{img}"
    dptImage = cv2.imread(path)
    height, width, _ = dptImage.shape
    dptImage = cv2.resize(dptImage, (rgbWidth, rgbHeight))

    cv2.imwrite(f"{destination_folder}{img}", dptImage)
```
