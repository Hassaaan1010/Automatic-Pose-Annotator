```python
import os
import cv2
import shutil
```

```python
#  define a source folder for images to prune
source_folder = "./dpt_images_3/"
dpt_list = os.listdir(source_folder)
dpt_list.sort(key=lambda x: int(x.split(".")[0]), reverse=True)
dpt_list[0:5]
```

```python
# create folders to match these paths before pruning.
ok_path = "./ok/"
delete_path = "./delete/"
ambigous_path = "./ambiguous/"
```

```python
dpt_list = os.listdir(source_folder)
dpt_list.sort(key=lambda x: int(x.split(".")[0]), reverse=True)

for image in dpt_list:
    print(image)
    path = f"{source_folder}{image}"

    frame = cv2.imread(path)

    while True:
        cv2.imshow(f"{image} | o:ok  d:del  q:ambiguous", frame)

        key = cv2.waitKey(0) & 0xFF  # Waits indefinitely for a key press

        if key == ord("d"):
            shutil.move(path, delete_path)
            break
        elif key == ord("o"):
            shutil.move(path, ok_path)
            break
        elif key == ord("q"):
            shutil.move(path, ambigous_path)
            break
    print(image, ":", chr(key))
    cv2.destroyAllWindows()  # Close the image window
```
